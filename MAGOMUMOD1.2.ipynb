Algunas variables presentan un número muy grande de categorías:

* **`Name of Exercise`:** 55 categorías
* **`Workout`:** 53 categorías

## **Columnas Sin Variabilidad**

* `meal_name` contiene **un único valor en todas las observaciones**.

## **Escalas Diferentes Entre Variables Numéricas**

Ejemplos:

* Variables fisiológicas: valores entre **0–200**
* Variables de calorías: valores entre **300–3000**
* Variables con datos muy grandes:

  * `Burns Calories (per 30 min)_bc` ≈ **10¹⁶–10²⁰**

**Recomendación:** aplicar **estandarización** (`StandardScaler`).

## **Problema**: `Burns Calories (per 30 min)_bc`

Esta variable muestra valores entre 10¹⁶–10²⁰, imposibles desde el punto de vista fisiológico para un gasto energético de una persona. Esto sugiere un error en la fórmula o en la generación del dato.

La variable **debería eliminarse**, ya que no aporta información válida y genera ruido en el dataset.

## **3.2. Tratamiento de Outliers**

Herramientas recomendadas:

* **Boxplots**
* **Z-score**
* **IQR**

Ejemplos detectados:

* **BMI** con valores superiores a 45
* **cal_balance** con valores extremadamente altos o negativos

```python
# Detección de outliers usando IQR
numeric_cols = df.select_dtypes(include=["int64", 'Float64']).columns
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1
outliers = ((df[numeric_cols] < (Q1 - 1.5 * IQR)) | (df[numeric_cols] > (Q3 + 1.5 * IQR)))
print("\nNúmero de outliers por columna:")
print(outliers.sum())
```

# 🏷️ **4. PREPROCESAMIENTO DE VARIABLES CATEGÓRICAS**

Las variables categóricas requieren tratamientos distintos según su estructura.


## **4.1. One-Hot Encoding**

Adecuada para variables **sin orden** jerárquico, como:

* `Gender`
* `Workout_Type`
* `diet_type`
* `cooking_method`
* `Body Part`

Convierte una columna con N categorías en N nuevas columnas binarias (con valores 0 o 1), una para cada categoría.

Así podemos representar estas categorías sin introducir un orden artificial.


```python
from sklearn.preprocessing import OneHotEncoder

nominal_cols = [
    "Gender",
    "Workout_Type",
    "diet_type",
    "cooking_method",
    "Body Part"
]

ohe = OneHotEncoder(sparse_output=False, drop=None, handle_unknown="ignore")

nominal_encoded = ohe.fit_transform(df_reduced[nominal_cols])
nominal_encoded_df = pd.DataFrame(
    nominal_encoded,
    columns=ohe.get_feature_names_out(nominal_cols)
)

nominal_encoded_df.head()
```

---



## **4.2. Label Encoding**

Para variables con **orden definido**, como:

* **`Difficulty Level`** (Easy < Intermediate < Hard)
* **`Experience_Level`** (ya numerada 1–3)

Asigna un número entero único a cada categoría. 




Ayuda a mantener el orden y facilitar la interpretación estadística.






```python
difficulty_mapping = {
    "Easy": 1,
    "Intermediate": 2,
    "Hard": 3
}

df_reduced["Difficulty_Level_ordinal"] = df_reduced["Difficulty Level"].map(difficulty_mapping)

df_reduced[["Difficulty Level", "Difficulty_Level_ordinal", "Experience_Level"]].head()
```

---


## **4.3. Codificación por Frecuencia**

Aplicada en variables con **muchas categorías**:

* `Name of Exercise`
* `Workout`
* `Target Muscle Group`

Nos permite reducir la dimensionalidad manteniendo la información útil.




```python
# Codificación por frecuencia de categorías para una variable de alta cardinalidad
col_high_card = "Name of Exercise"

freq_map = df_reduced[col_high_card].value_counts(normalize=True).to_dict()
df_reduced[col_high_card + "_freq"] = df_reduced[col_high_card].map(freq_map)

df_reduced[[col_high_card, col_high_card + "_freq"]].head()
```
